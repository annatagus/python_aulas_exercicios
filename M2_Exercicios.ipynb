## GRUPO A: Listas

### 1. Lista de Disciplinas

```
# Criar a lista com pelo menos 4 disciplinas
disciplinas = ["Matemática", "Programação", "Bases de Dados", "Estatística"]

# a) Lista completa
print("a) Lista completa:", disciplinas)

# b) Primeiro elemento
print("b) Primeiro elemento:", disciplinas[0])

# c) Último elemento
print("c) Último elemento:", disciplinas[-1])
```

### 2. Lista com Tipos Diferentes

```
# Lista com inteiros, floats, strings e booleanos
valores = [10, "Python", 3.14, True, [1, 2]]

# Percorrer e mostrar o tipo de dados com type()
for elemento in valores:
    print(f"O valor '{elemento}' é do tipo: {type(elemento)}")
```

### 3. Elementos Duplicados e Tamanho

```
# Lista com elementos duplicados
numeros = [1, 2, 2, 3, 4, 4, 4, 5]

# len() contabiliza todos os elementos, incluindo duplicados
print("Lista:", numeros)
print("Número total de elementos (len):", len(numeros))
```

### 4. Matriz 2x3 e Acesso a Índices

```
# Criar matriz 2x3
matriz = [
    [10, 20, 30],
    [40, 50, 60]
]

print("Matriz 2x3 completa:", matriz)

# Aceder ao valor 50 (linha 1, coluna 1)
valor_50 = matriz[1][1]
print("O valor acedido na posição [1][1] é:", valor_50)
```

### 5. Soma de Matriz 3x3

```
matriz_3x3 = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

soma_total = 0

# Dois ciclos for encaixados para somar todos os elementos
for linha in matriz_3x3:
    for elemento in linha:
        soma_total += elemento

print("A soma de todos os elementos da matriz 3x3 é:", soma_total)
```

### 6. Operações com Lista de Números

```
lista_original = [1, 2, 3, 4, 5, 6]

# a) Só números pares usando for
pares_for = []
for num in lista_original:
    if num % 2 == 0:
        pares_for.append(num)
print("a) Pares com for:", pares_for)

# b) Só números pares usando list comprehension
pares_lc = [num for num in lista_original if num % 2 == 0]
print("b) Pares com list comprehension:", pares_lc)

# c) Cubo dos números maiores que 3
cubos = [num ** 3 for num in lista_original if num > 3]
print("c) Cubo dos números > 3:", cubos)
```

### 7. Verificação, Ordenação e Reatribuição

```
# Verificação de existência
lista_a = [2, 4, 6, 8]
print("10 existe em [2, 4, 6, 8]?", 10 in lista_a)

# Ordenação crescente e decrescente
lista_b = [5, 1, 4, 2]
lista_b.sort()
print("Ordem crescente:", lista_b)

lista_b.sort(reverse=True)
print("Ordem decrescente:", lista_b)

# Reatribuição do último elemento para 0
lista_b[-1] = 0
print("Após reatribuir o último elemento para 0:", lista_b)
```

### 8. Remoção de Elementos

```
numeros = [1, 2, 3, 4, 5]

# a) Remover o valor 3 com remove()
numeros.remove(3)
print("a) Após remove(3):", numeros)

# b) Remover o primeiro elemento com del
del numeros[0]
print("b) Após del numeros[0]:", numeros)

# c) Usar pop() e mostrar valor removido e lista final
removido = numeros.pop()
print(f"c) Valor removido com pop(): {removido} | Lista final: {numeros}")
```

### 9. Alteração de Lista

```
elementos = ["A", "B", "C", "D", "E"]

# Mostrar número de elementos, primeiro e último
print("Número de elementos:", len(elementos))
print("Primeiro elemento:", elementos[0])
print("Último elemento:", elementos[-1])

# Adicionar valor com append()
elementos.append("F")

# Alterar posição 2 para 100
elementos[2] = 100

print("Lista atualizada:", elementos)
```

## GRUPO C: Dicionários

### 13. Menu de Gestão de Aluno

```
aluno = {
    "numero": 101,
    "nome": "João Silva",
    "curso": "Data Analytics",
    "nota_final": 14.5
}

opcao = ""
while opcao != "d":
    print("\n--- MENU ALUNO ---")
    print("a) Mostrar nome e nota")
    print("b) Atualizar a nota")
    print("c) Adicionar 'aprovado'")
    print("d) Sair")

    opcao = input("Escolha uma opção: ").strip().lower()

    if opcao == "a":
        print(f"Nome: {aluno['nome']} | Nota Final: {aluno['nota_final']}")
    elif opcao == "b":
        nova_nota = float(input("Digite a nova nota: "))
        aluno["nota_final"] = nova_nota
        print("Nota atualizada com sucesso!")
    elif opcao == "c":
        aluno["aprovado"] = aluno["nota_final"] >= 10
        print(f"Estado 'aprovado' adicionado/atualizado para: {aluno['aprovado']}")
    elif opcao == "d":
        print("A sair do programa...")
    else:
        print("Opção inválida! Tente novamente.")
```

### 14. Iteração sobre Dicionário

```
produto = {
    "codigo": "P001",
    "descricao": "Teclado Mecânico",
    "preco": 59.99,
    "stock": 15
}

# a) Todas as chaves
print("a) Chaves:")
for chave in produto.keys():
    print(chave)

# b) Todos os valores
print("\nb) Valores:")
for valor in produto.values():
    print(valor)

# c) Par chave: valor
print("\nc) Par chave: valor:")
for chave, valor in produto.items():
    print(f"{chave}: {valor}")
```

## GRUPO D: Sets

### 16. Operações num Set

```
alunos_inscritos = {"Ana", "Bruno", "Carlos"}

# a) Adicionar um novo aluno
alunos_inscritos.add("Daniela")

# b) Remover um aluno
alunos_inscritos.remove("Carlos")

# c) Verificar se 'João' está inscrito
joao_inscrito = "João" in alunos_inscritos
print("João está inscrito?", joao_inscrito)

# d) Imprimir set final
print("Set final de inscritos:", alunos_inscritos)
```

### 17. Operações entre Sets

```
disciplinas_ana = {"Matemática", "Física", "Programação"}
disciplinas_bruno = {"Programação", "Química", "História", "Matemática"}

# a) Interseção (em ambas)
intersecao = disciplinas_ana.intersection(disciplinas_bruno)
print("a) Interseção:", intersecao)

# b) Diferença (apenas de Ana)
diferenca = disciplinas_ana.difference(disciplinas_bruno)
print("b) Diferença (Ana - Bruno):", diferenca)

# c) União (todas as disciplinas sem repetição)
uniao = disciplinas_ana.union(disciplinas_bruno)
print("c) União:", uniao)
```

### 18. Processamento de Frase com Set

```
frase = "python é simples e python é poderoso"

# a) Converter numa lista de palavras usando split()
palavras_lista = frase.split()
print("a) Lista de palavras:", palavras_lista)

# b) Crie um set com as palavras únicas
palavras_unicas = set(palavras_lista)

# c) Mostre o set e o número de palavras únicas
print("c) Set de palavras únicas:", palavras_unicas)
print("   Número de palavras únicas:", len(palavras_unicas))
```

## GRUPO E: Integração e Menus

### 19. Menu To-Do List (com Lista)

```
tarefas = []
opcao = ""

while opcao != "4":
    print("\n--- TO-DO LIST ---")
    print("1. Adicionar tarefa")
    print("2. Listar tarefas")
    print("3. Remover tarefa pelo nome")
    print("4. Sair")

    opcao = input("Opção: ").strip()

    if opcao == "1":
        nova_tarefa = input("Digite a nova tarefa: ").strip()
        if nova_tarefa:
            tarefas.append(nova_tarefa)
            print("Tarefa adicionada!")
    elif opcao == "2":
        print("\nLista de Tarefas:")
        if not tarefas:
            print("(Lista vazia)")
        else:
            for idx, t in enumerate(tarefas, start=1):
                print(f"{idx}. {t}")
    elif opcao == "3":
        remover = input("Digite o nome exato da tarefa a remover: ").strip()
        if remover in tarefas:
            tarefas.remove(remover)
            print("Tarefa removida!")
        else:
            print("Tarefa não encontrada.")
    elif opcao == "4":
        print("Saindo...")
    else:
        print("Opção inválida!")
```

### 20. Menu Gestão de Alunos (com Lista)

```
alunos = []
opcao = ""

while opcao != "4":
    print("\n--- GESTÃO DE ALUNOS ---")
    print("1. Adicionar aluno")
    print("2. Remover aluno pelo nome")
    print("3. Listar todos os alunos")
    print("4. Sair")

    opcao = input("Opção: ").strip()

    if opcao == "1":
        nome = input("Digite o nome do aluno: ").strip()
        if nome:
            alunos.append(nome)
            print("Aluno adicionado!")
    elif opcao == "2":
        nome = input("Digite o nome do aluno a remover: ").strip()
        if nome in alunos:
            alunos.remove(nome)
            print("Aluno removido!")
        else:
            print("Aluno não encontrado.")
    elif opcao == "3":
        print("\nLista de Alunos:")
        if not alunos:
            print("(Nenhum aluno registado)")
        else:
            for a in alunos:
                print("-", a)
    elif opcao == "4":
        print("Saindo...")
    else:
        print("Opção inválida!")
```

### 21. Menu Gestor de Passwords (com Dicionário)

```
passwords = {}  # {utilizador: password}
opcao = ""

while opcao != "4":
    print("\n--- GESTOR DE PASSWORDS ---")
    print("1. Adicionar utilizador e password")
    print("2. Ver password de um utilizador")
    print("3. Apagar utilizador")
    print("4. Sair")

    opcao = input("Opção: ").strip()

    if opcao == "1":
        user = input("Nome de utilizador: ").strip()
        pwd = input("Password: ").strip()
        passwords[user] = pwd
        print("Utilizador registado/atualizado!")
    elif opcao == "2":
        user = input("Digite o nome do utilizador: ").strip()
        if user in passwords:
            print(f"Password de {user}: {passwords[user]}")
        else:
            print("Utilizador não encontrado.")
    elif opcao == "3":
        user = input("Digite o nome do utilizador a apagar: ").strip()
        if user in passwords:
            del passwords[user]
            print("Utilizador apagado!")
        else:
            print("Utilizador não encontrado.")
    elif opcao == "4":
        print("Saindo...")
    else:
        print("Opção inválida!")
```

### 22. Reimplementação da Gestão de Alunos usando `set` (Justificação Incluída)

```
# Reimplementação do Exercício 20 utilizando Set
alunos_set = set()
opcao = ""

while opcao != "4":
    print("\n--- GESTÃO DE ALUNOS (USANDO SET) ---")
    print("1. Adicionar aluno")
    print("2. Remover aluno pelo nome")
    print("3. Listar todos os alunos")
    print("4. Sair")

    opcao = input("Opção: ").strip()

    if opcao == "1":
        nome = input("Digite o nome do aluno: ").strip()
        if nome:
            alunos_set.add(nome)
            print("Aluno adicionado!")
    elif opcao == "2":
        nome = input("Digite o nome do aluno a remover: ").strip()
        if nome in alunos_set:
            alunos_set.remove(nome)
            print("Aluno removido!")
        else:
            print("Aluno não encontrado.")
    elif opcao == "3":
        print("\nLista de Alunos:")
        if not alunos_set:
            print("(Nenhum aluno registado)")
        else:
            for a in alunos_set:
                print("-", a)
    elif opcao == "4":
        print("Saindo...")
    else:
        print("Opção inválida!")

# JUSTIFICAÇÃO:
# Escolheu-se a estrutura 'set' em vez de 'list' porque o set impede automaticamente
# a inserção de nomes duplicados na turma e oferece uma pesquisa e remoção mais rápida.
```

### 23. Dicionário Complexo de Turma e Consultas

```
# Estrutura do dicionário para gerir uma turma
turma = {
    101: {
        "nome": "Ana Costa",
        "disciplinas": ["Matemática", "Programação", "Inglês"]
    },
    102: {
        "nome": "Bernardo Dias",
        "disciplinas": ["Bases de Dados", "Programação"]
    }
}

opcao = ""
while opcao != "3":
    print("\n--- CONSULTA DE TURMA ---")
    print("1. Listar todos os alunos e disciplinas")
    print("2. Consultar disciplinas de um aluno pelo número")
    print("3. Sair")

    opcao = input("Opção: ").strip()

    if opcao == "1":
        for numero, dados in turma.items():
            print(f"Número: {numero} | Nome: {dados['nome']}")
            print(f"   Disciplinas: {', '.join(dados['disciplinas'])}")
    elif opcao == "2":
        num_pesquisa = int(input("Digite o número do aluno: "))
        if num_pesquisa in turma:
            aluno_info = turma[num_pesquisa]
            print(f"Aluno: {aluno_info['nome']}")
            print("Disciplinas inscritas:")
            for disc in aluno_info["disciplinas"]:
                print(f"  - {disc}")
        else:
            print("Aluno não encontrado.")
    elif opcao == "3":
        print("Saindo do sistema de turma...")
    else:
        print("Opção inválida!")
```
