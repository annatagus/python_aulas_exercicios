```python
# !pip -q install pyspark
from pyspark.sql import SparkSession
spark = (SparkSession.builder
         .appName("Spark-Exercises-Colab")
         .getOrCreate())
spark
```

```python
from pyspark.sql import functions as F
from pyspark.sql import types as T
```

```python
customers_data = [
  (1001, "Ana",   "PT", "2024-07-01", "gold"),
  (1002, "Luis",  "ES", "2024-09-15", "silver"),
  (1003, "Marta", "PT", "2025-01-01", "bronze"),
  (1004, "João",  "PT", "2025-01-05", "silver"),
  (1005, "John",  "US", "2024-12-20", "gold"),
]
```

```python
customers_schema = T.StructType([
  T.StructField("customer_id", T.IntegerType(), False),
  T.StructField("customer_name", T.StringType(), False),
  T.StructField("country", T.StringType(), False),
  T.StructField("signup_date", T.StringType(), False),
  T.StructField("tier", T.StringType(), False),
])
```

```python
customers = spark.createDataFrame(customers_data, schema=customers_schema).withColumn("signup_date", F.to_date("signup_date"))
```

```python
customers.createOrReplaceTempView("customers")
spark.sql("""
    SELECT *
    FROM customers
""").show()
```

```python
customers.createOrReplaceTempView("customers")
spark.sql("""
    SELECT *
    FROM customers
    WHERE country = 'PT'
""").show()
```

```python
customers.createOrReplaceTempView("customers")
spark.sql("""
    SELECT COUNT(1) AS num_c
    FROM customers
    WHERE country = 'PT'
""").show()
```

```python
customers.createOrReplaceTempView("customersV2")
spark.sql("""
    SELECT country, COUNT(1) as num_c
    FROM customers
    GROUP BY country
""").show()
```

```python
customers.show()
```

```python
customers.where((F.col("country") == "PT") & (F.col("tier") == "gold")).show()
```
