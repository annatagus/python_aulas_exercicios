## GRUPO A: Dataset Iris

### Importação de Bibliotecas e Carregamento do Dataset

```
import pandas as pd
import numpy as np
import seaborn as sns

# Carregar o dataset Iris
df_iris = sns.load_dataset("iris")
```

### Exercício 1. Exploração Inicial

```
# 1. Mostrar as primeiras 5 linhas
print("--- Primeiras 5 linhas ---")
print(df_iris.head())

# 2. Número de linhas e de colunas
print("\n--- Dimensões (Linhas, Colunas) ---")
print(df_iris.shape)

# 3. Tipos de dados de cada coluna
print("\n--- Tipos de dados ---")
print(df_iris.dtypes)

# 4. Verificar valores nulos
print("\n--- Contagem de valores nulos ---")
#print(df_iris.isnull())
print(df_iris.isnull().sum())

# 5. Definir a coluna 'species' como índice
df_iris_idx = df_iris.set_index("species")
print("\n--- DataFrame com 'species' como índice ---")
print(df_iris_idx.head())
```

### Exercício 2. Seleção e Filtragem

```
# 1. Selecionar apenas 'sepal_length', 'petal_length' e 'species'
df_sub = df_iris[["sepal_length", "petal_length", "species"]]
print("--- Subconjunto de colunas ---")
print(df_sub.head())

# 2. Filtrar apenas registos da espécie 'setosa'
df_setosa = df_iris[df_iris["species"] == "setosa"]
print("\n--- Registos da espécie Setosa ---")
print(df_setosa.head())

# 3. Calcular o comprimento médio da pétala dessa espécie
media_petala_setosa = df_setosa["petal_length"].mean()
print(f"\nComprimento médio da pétala (Setosa): {media_petala_setosa:.2f} cm")

# 4. Exportar o resultado para CSV 'setosa.csv'
df_setosa.to_csv("setosa.csv", index=False)
print("Ficheiro 'setosa.csv' exportado com sucesso.")
```

### Exercício 3. Agregações Simples

```
# 1. Média de todas as variáveis numéricas
print("--- Média de todas as variáveis numéricas ---")
print(df_iris.mean(numeric_only=True))

# 2. Valor máximo e mínimo de 'petal_length'
max_petal = df_iris["petal_length"].max()
min_petal = df_iris["petal_length"].min()
print(f"\nPetal Length -> Mínimo: {min_petal} | Máximo: {max_petal}")

# 3. Desvio padrão de 'sepal_width'
std_sepal_w = df_iris["sepal_width"].std()
print(f"\nDesvio Padrão de 'sepal_width': {std_sepal_w:.2f}")

# 4. Método describe() e interpretação
print("\n--- Resumo Estatístico (describe) ---")
print(df_iris.describe())

# INTERPRETAÇÃO:
# O describe() mostra que as pétalas variam significativamente mais (desvio padrão de 1.76 cm no petal_length)
# do que a largura da sépala (desvio padrão de 0.44 cm). A mediana (50%) do comprimento da pétala é 4.35 cm,
# enquanto a média é 3.76 cm, indicando a presença de agrupamentos de valores mais pequenos (espécie Setosa).
```

### Exercício 4. Agrupamento e Agregação (groupby)

```
# 1. Média de 'petal_length' por espécie
print("--- Média de 'petal_length' por espécie ---")
print(df_iris.groupby("species")["petal_length"].mean())

# 2. Média, mín, máx e desvio padrão com agg()
print("\n--- Múltiplas agregações de 'petal_length' por espécie ---")
agg_petal = df_iris.groupby("species")["petal_length"].agg(["mean", "min", "max", "std"])
print(agg_petal)

# 3. Espécie com maior comprimento médio da pétala
especie_maior = agg_petal["mean"].idxmax()
print(f"\nA espécie com maior comprimento médio de pétala é: {especie_maior}")

# 4. Ordenar os resultados por ordem decrescente da média
agg_petal_ord = agg_petal.sort_values(by="mean", ascending=False)
print("\n--- Resultados ordenados por média decrescente ---")
print(agg_petal_ord)

# 5. Resetar o índice após o agrupamento
df_resumo = agg_petal_ord.reset_index()
print("\n--- DataFrame com índice resetado ---")
print(df_resumo)

# 6. Agregação personalizada: Intervalo (máx - mín) por espécie
df_intervalo = df_iris.groupby("species")["petal_length"].apply(lambda x: x.max() - x.min()).reset_index()
df_intervalo.columns = ["species", "amplitude_petal_length"]
print("\n--- Intervalo (Amplitude) de petal_length por espécie ---")
print(df_intervalo)
```

## GRUPO B: Dataset Tips

### Exercício 5. Análise de Gorjetas

```
# Carregar o dataset Tips
df_tips = sns.load_dataset("tips")

# PARTE A: Exploração inicial
print("--- Primeiras 5 linhas ---")
print(df_tips.head())

print("\n--- Dimensões e Tipos de Dados ---")
print(df_tips.info())

print("\n--- Valores Nulos ---")
print(df_tips.isnull().sum())

# PARTE B: Criação de variável derivada
# Percentagem da gorjeta em relação ao total da conta
df_tips["tip_percent"] = (df_tips["tip"] / df_tips["total_bill"]) * 100
print("\n--- Primeiras linhas com 'tip_percent' ---")
print(df_tips[["total_bill", "tip", "tip_percent"]].head())

# PARTE C: Agregações
media_conta = df_tips["total_bill"].mean()
media_gorjeta = df_tips["tip"].mean()
media_perc = df_tips["tip_percent"].mean()
std_perc = df_tips["tip_percent"].std()

print(f"\nValor médio da conta: {media_conta:.2f} $")
print(f"Valor médio da gorjeta: {media_gorjeta:.2f} $")
print(f"Percentagem média da gorjeta: {media_perc:.2f}%")
print(f"Desvio padrão da percentagem: {std_perc:.2f}%")

# PARTE D: Agrupamento e comparação
print("\n--- Percentagem média por dia e sexo ---")
grp_dia_sexo = df_tips.groupby(["day", "sex"], observed=False)["tip_percent"].mean().reset_index()
print(grp_dia_sexo)

print("\n--- Média da gorjeta e da percentagem agrupado por dia e sexo ---")
grp_completo = df_tips.groupby(["day", "sex"], observed=False).agg(
    media_gorjeta=("tip", "mean"),
    media_percentagem=("tip_percent", "mean")
).reset_index()

# Ordenar pela maior percentagem média
grp_ordenado = grp_completo.sort_values(by="media_percentagem", ascending=False)
print("\n--- Resultados ordenados por maior percentagem média ---")
print(grp_ordenado)

# PARTE E: Interpretação e Conclusões
# 11. Dia com maior percentagem média
dia_maior_perc = df_tips.groupby("day", observed=False)["tip_percent"].mean().idxmax()
print(f"\nO dia com maior percentagem média de gorjeta é: {dia_maior_perc}")

# 12 & 13. Conclusões fundamentadas:
# - Conclusão 1: Ao Domingo (Sun) e Sexta (Fri), os clientes tendem a dar uma percentagem ligeiramente superior de gorjeta em média.
# - Conclusão 2: As mulheres deixam, em média, uma percentagem de gorjeta ligeiramente superior (ou equivalente) à dos homens, apesar de o valor absoluto da conta (total_bill) dos homens ser tendencialmente maior.
# - Conclusão 3: Existe uma grande variabilidade na percentagem de gorjetas em contas de menor valor, onde gorjetas fixas (ex: 2$) inflamam significativamente o valor do 'tip_percent'.

# PARTE F: Exportação
grp_ordenado.to_csv("tips_analysis.csv", index=False)
print("\nFicheiro 'tips_analysis.csv' exportado com sucesso.")
```

## GRUPO C: Dataset Titanic

### Exercício 6. Análise de Sobrevivência

```
# Carregar o dataset Titanic
df_titanic = sns.load_dataset("titanic")

# 1. Taxa global de sobrevivência (a coluna 'survived' é 0 ou 1)
taxa_global = df_titanic["survived"].mean() * 100
print(f"1. Taxa global de sobrevivência: {taxa_global:.2f}%")

# 2. Taxa de sobrevivência por sexo
taxa_sexo = df_titanic.groupby("sex")["survived"].mean() * 100
print("\n2. Taxa de sobrevivência por sexo:")
print(taxa_sexo)

# 3. Taxa de sobrevivência por classe (pclass)
taxa_classe = df_titanic.groupby("pclass")["survived"].mean() * 100
print("\n3. Taxa de sobrevivência por classe:")
print(taxa_classe)

# 4. Agrupar por sexo e classe simultaneamente
taxa_grupo = df_titanic.groupby(["sex", "pclass"])["survived"].mean().reset_index()
taxa_grupo["survived_percent"] = taxa_grupo["survived"] * 100
print("\n4. Taxa de sobrevivência por sexo e classe:")
print(taxa_grupo)

# 5. Identificar o grupo com maior taxa de sobrevivência
grupo_top = taxa_grupo.loc[taxa_grupo["survived_percent"].idxmax()]
print(f"\n5. O grupo com maior taxa de sobrevivência foi: {grupo_top['sex']} na classe {grupo_top['pclass']} ({grupo_top['survived_percent']:.2f}%).")
```

## GRUPO D: Consolidação (Dataset à Escolha)

### Exercício 7. Projeto de Consolidação (Dataset: `penguins`)

```
# Escolha do dataset do Seaborn: Penguins
df_penguins = sns.load_dataset("penguins")

# 1. Exploração inicial
print("--- 1. Exploração Inicial ---")
print(df_penguins.head())
print("\nDimensões:", df_penguins.shape)
print("\nValores nulos por coluna:\n", df_penguins.isnull().sum())

# Tratar os nulos para análise rigorosa
df_clean = df_penguins.dropna()

# 2. Dois filtros
# Filtro 1: Pinguins da ilha Biscoe
filtro_biscoe = df_clean[df_clean["island"] == "Biscoe"]

# Filtro 2: Pinguins com massa corporal superior a 4000g
filtro_pesados = df_clean[df_clean["body_mass_g"] > 4000]

print(f"\nNº de pinguins na ilha Biscoe: {len(filtro_biscoe)}")
print(f"Nº de pinguins com massa > 4000g: {len(filtro_pesados)}")

# 3. Duas agregações simples
massa_media = df_clean["body_mass_g"].mean()
bico_max = df_clean["bill_length_mm"].max()
print(f"\nMassa corporal média: {massa_media:.2f} g")
print(f"Comprimento máximo do bico: {bico_max} mm")

# 4. Um agrupamento com múltiplas agregações
# Agrupar por espécie e sexo, calculando a média e desvio padrão do peso e tamanho da barbatana
analise_especie = df_clean.groupby(["species", "sex"]).agg(
    massa_media=("body_mass_g", "mean"),
    massa_std=("body_mass_g", "std"),
    barbatana_media=("flipper_length_mm", "mean")
).reset_index()

print("\n--- Agrupamento por Espécie e Sexo ---")
print(analise_especie)

# 5. Exportação para CSV
analise_especie.to_csv("penguins_analysis.csv", index=False)
print("\nResultado exportado para 'penguins_analysis.csv'.")

# 6. Três conclusões fundamentadas com base nos dados:
# - Conclusão 1: A espécie Gentoo é significativamente mais pesada que as espécies Adelie e Chinstrap, com uma massa média superior a 5000g nos machos.
# - Conclusão 2: Em todas as três espécies, os machos apresentam uma massa corporal e um comprimento de barbatana superior ao das fêmeas (dimorfismo sexual).
# - Conclusão 3: A variação de peso (desvio padrão) é proporcionalmente menor na espécie Chinstrap, indicando uma maior homogeneidade de tamanho nesta população.
```
