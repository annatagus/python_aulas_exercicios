## GRUPO A: Sequência, Variáveis e Operadores Aritméticos

### 1. Média de Notas

```
# Leitura das notas
nota_pt = float(input("Digite a nota de Português: "))
nota_ing = float(input("Digite a nota de Inglês: "))
nota_mat = float(input("Digite a nota de Matemática: "))

# Cálculo da média aritmética
media = (nota_pt + nota_ing + nota_mat) / 3

# Apresentação do resultado formatado com 2 casas decimais
print(f"A média aritmética é: {media:.2f}")
```

### 2. Desconto de 30%

```
preco = float(input("Digite o preço original do produto (€): "))
# Aplicar 30% de desconto (preço * 0.70)
preco_com_desconto = preco * 0.70

print(f"O preço final com 30% de desconto é: {preco_com_desconto:.2f} €")
```

### 3. Área de um Círculo

```
raio = float(input("Digite o valor do raio: "))
pi = 3.14159

# Cálculo da área usando o operador **
area = pi * (raio ** 2)

print(f"A área do círculo é: {area:.2f}")
```

### 4. Anos até aos 100

```
idade = int(input("Digite a sua idade: "))
anos_faltam = 100 - idade

print(f"Faltam {anos_faltam} anos para completar 100 anos.")
```

### 5. Converter Euros para Cêntimos

```
euros = float(input("Digite o valor em euros (€): "))
centimos = int(euros * 100)

print(f"{euros:.2f}€ equivalem a {centimos} cêntimos.")
```

### 6. Perímetro de um Retângulo

```
largura = float(input("Digite a largura do retângulo: "))
altura = float(input("Digite a altura do retângulo: "))

perimetro = 2 * (largura + altura)

print(f"O perímetro do retângulo é: {perimetro}")
```

### 7. Dobro e Triplo

```
numero = int(input("Digite um número inteiro: "))

dobro = numero * 2
triplo = numero * 3

print(f"O dobro é {dobro} e o triplo é {triplo}.")
```

## GRUPO B: Operadores Relacionais e Lógicos

### 1. Dar Sangue (entre 18 e 67 anos)

```
idade = int(input("Digite a sua idade: "))

# Combinação de duas comparações com o operador and
pode_doar = idade >= 18 and idade <= 67

print(f"Pode doar sangue? {pode_doar}")
```

### 2. Condição para Conduzir

```
altura = float(input("Digite a sua altura em metros: "))
idade = int(input("Digite a sua idade: "))
tem_carta = input("Tem carta de condução? (s/n): ").strip().lower() == 's'

# Pelo menos 1.5m de altura E (22 anos ou mais OU carta de condução)
pode_conduzir = altura >= 1.5 and (idade >= 22 or tem_carta)

print(f"Consegue conduzir? {pode_conduzir}")
```

### 3. Verificar se é Vogal

### 4. Entrada na Discoteca

### 5. Dois Números Positivos

```
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))

ambos_positivos = (num1 > 0) and (num2 > 0)

print(f"Ambos os números são positivos? {ambos_positivos}")
```

### 6. Expressão Lógica Composta

```
# Definição de três valores lógicos
a = True
b = False
c = True

# Exemplo de expressão lógica composta
resultado = (a or b) and not c

print(f"Resultado da expressão: {resultado}")
```

### 7. Positivo e Par

```
numero = int(input("Digite um número: "))

# É estritamente positivo (> 0) e o resto da divisão por 2 é 0
e_positivo_e_par = (numero > 0) and (numero % 2 == 0)

print(f"O número é simultaneamente positivo e par? {e_positivo_e_par}")
```

## GRUPO C: Estruturas Condicionais (IF / ELIF / ELSE)

### 1. Positivo, Negativo ou Zero

```
numero = int(input("Digite um número inteiro: "))

if numero > 0:
    print("O número é positivo.")
elif numero < 0:
    print("O número é negativo.")
else:
    print("O número é zero.")
```

### 2. Aprovado ou Reprovado

```
nota = float(input("Digite a nota do aluno: "))

if nota >= 10:
    print("Aprovado")
else:
    print("Reprovado")
```

### 3. Maior e Menor Número

```
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))

if num1 > num2:
    print(f"O maior número é {num1} e o menor é {num2}.")
elif num2 > num1:
    print(f"O maior número é {num2} e o menor é {num1}.")
else:
    print("Ambos os números são iguais.")
```

### 4. Par ou Ímpar

```
numero = int(input("Digite um número inteiro: "))

if numero % 2 == 0:
    print("O número é par.")
else:
    print("O número é ímpar.")
```

### 5. Escala de Idade

```
idade = int(input("Digite a idade: "))

if idade < 18:
    print("Menor de idade")
elif idade < 65:
    print("Adulto")
else:
    print("Idoso")
```

### 6. Estado da Temperatura

```
temp = float(input("Digite a temperatura (°C): "))

if temp < 15:
    print("Está frio.")
elif temp <= 25:
    print("Está ameno.")
else:
    print("Está quente.")
```

### 7. Intervalo entre 10 e 20

## GRUPO D: Estruturas Iterativas - Ciclo FOR

### 1. Números Pares de 1 a 20

```
# range(2, 21, 2) gera os números de 2 até 20 de 2 em 2
for num in range(2, 21, 2):
    print(num)
```

### 2. Imprimir Nomes da Lista

```
alunos = ["Ana", "Bernardo", "Catarina", "Diogo"]

for nome in alunos:
    print(nome)
```

### 3. Contar Ocorrências de um Carácter

```
texto = input("Digite uma palavra ou frase: ")
alvo = input("Digite o carácter que pretende contar: ")

contador = 0
for caracter in texto:
    if caracter == alvo:
        contador += 1

print(f"O carácter '{alvo}' aparece {contador} vez(es).")
```

### 4. Múltiplos de 3 entre 1 e 30

```
for num in range(1, 31):
    if num % 3 == 0:
        print(num)
```

### 5. Soma de 1 a 100

```
soma = 0
for i in range(1, 101):
    soma += i

print(f"A soma de todos os números de 1 a 100 é: {soma}")
```

### 6. Maior Valor de uma Lista

```
numeros = [14, 52, 3, 98, 27, 41]

# Assumir que o primeiro elemento é o maior
maior = numeros[0]

for num in numeros:
    if num > maior:
        maior = num

print(f"O maior valor da lista é: {maior}")
```

### 7. Tabuada

```
numero = int(input("Digite um número para calcular a tabuada: "))

for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
```

## GRUPO E: Estruturas Iterativas - Ciclo WHILE

### 1. Soma Superior a 100

```
soma = 0

while soma <= 100:
    numero = float(input(f"Soma atual = {soma}. Digite um número: "))
    soma += numero

print(f"Soma final: {soma} (ultrapassou 100).")
```

### 2. Validação de Password

```
password_correta = "python123"
pass_introduzida = input("Digite a password: ")

while pass_introduzida != password_correta:
    print("Password incorreta! Tente novamente.")
    pass_introduzida = input("Digite a password: ")

print("Acesso concedido!")
```

### 3. Ler até Parar no 0

```
numero = float(input("Digite um número (0 para sair): "))

while numero != 0:
    print(f"Valor introduzido: {numero}")
    numero = float(input("Digite outro número (0 para sair): "))

print("Terminado.")
```

### 4. Contar Números Introduzidos

```
contador = 0
opcao = input("Deseja introduzir um número? (s/n): ").strip().lower()

while opcao == 's':
    input("Digite o número: ")
    contador += 1
    opcao = input("Deseja introduzir outro número? (s/n): ").strip().lower()

print(f"Total de números introduzidos: {contador}")
```

### 5. Média de Notas Válidas (0 a 20)

```
soma_notas = 0
contador = 0

nota = float(input("Digite uma nota (0 a 20): "))

# O ciclo continua enquanto a nota estiver no intervalo válido
while 0 <= nota <= 20:
    soma_notas += nota
    contador += 1
    nota = float(input("Digite a nota seguinte (ou valor fora de 0-20 para terminar): "))

if contador > 0:
    media = soma_notas / contador
    print(f"Média das {contador} notas válidas: {media:.2f}")
else:
    print("Nenhuma nota válida foi inserida.")
```

### 6. Contagem Decrescente de 20 a 1

```
numero = 20

while numero >= 1:
    print(numero)
    numero -= 1
```

## GRUPO F: Desafios (Consolidação)

### 1. Fatorial de um Número

```
numero = int(input("Digite um número inteiro não-negativo: "))
fatorial = 1

for i in range(1, numero + 1):
    fatorial *= i

print(f"O fatorial de {numero} ({numero}!) é: {fatorial}")
```

### 2. Potência sem Operador `**`

```
base = float(input("Digite a base: "))
expoente = int(input("Digite o expoente (inteiro >= 0): "))

resultado = 1
for _ in range(expoente):
    resultado *= base

print(f"{base} elevado a {expoente} é: {resultado}")
```

### 3. Contagem de Positivos, Negativos e Zeros

```
positivos = 0
negativos = 0
zeros = 0

quantidade = int(input("Quantos números pretende introduzir? "))

for i in range(quantidade):
    num = float(input(f"Número {i + 1}: "))
    if num > 0:
        positivos += 1
    elif num < 0:
        negativos += 1
    else:
        zeros += 1

print(f"Positivos: {positivos} | Negativos: {negativos} | Zeros: {zeros}")
```

### 4. Contar Vogais numa String

```
texto = input("Digite uma palavra ou frase: ")
vogais = "aeiouAEIOU"
contador = 0

for caracter in texto:
    if caracter in vogais:
        contador += 1

print(f"A frase contém {contador} vogal(ais).")
```
