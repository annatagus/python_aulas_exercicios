```
def quadrado(numero):
    return numero ** 2

# Guardar o valor devolvido numa variável e imprimir
resultado = quadrado(5)
print("O quadrado de 5 é:", resultado)
```

### 4. Área do Retângulo (Argumentos Posicionais vs. Nomeados)

```
def calcular_area(largura, altura):
    return largura * altura

# Chamada com argumentos posicionais
area_posicional = calcular_area(5, 10)
print("Área (posicional):", area_posicional)

# Chamada com argumentos nomeados
area_nomeada = calcular_area(altura=10, largura=5)
print("Área (nomeada):", area_nomeada)
```

### 5. Refactoring com Funções

#### a) Eliminar código repetido de soma

```
# Função para evitar repetição da operação de soma
def soma(a, b):
    return a + b

# Utilização da função
res1 = soma(10, 20)
res2 = soma(5.5, 4.5)

print(f"Resultado 1: {res1}")
print(f"Resultado 2: {res2}")
```

#### b) Converter Euros para várias moedas

```
def converter_euro(valor_euro, taxa_cambio):
    return valor_euro * taxa_cambio

euros = 100

# Taxas de exemplo
usd = converter_euro(euros, 1.08)
gbp = converter_euro(euros, 0.85)
brl = converter_euro(euros, 5.50)

print(f"{euros}€ equivalem a {usd:.2f} USD")
print(f"{euros}€ equivalem a {gbp:.2f} GBP")
print(f"{euros}€ equivalem a {brl:.2f} BRL")
```

### 6. Calculadora com Funções e Menu

```
# Operações fundamentais
def soma(a, b):
    return a + b

def subtracao(a, b):
    return a - b

def multiplicacao(a, b):
    return a * b

def divisao(a, b):
    if b == 0:
        return "Erro: Divisão por zero!"
    return a / b

# Programa Principal (Menu)
opcao = ""
while opcao != "5":
    print("\n--- CALCULADORA ---")
    print("1. Soma")
    print("2. Subtração")
    print("3. Multiplicação")
    print("4. Divisão")
    print("5. Sair")

    opcao = input("Escolha uma opção: ").strip()

    if opcao in ["1", "2", "3", "4"]:
        num1 = float(input("Digite o primeiro número: "))
        num2 = float(input("Digite o segundo número: "))

        if opcao == "1":
            print(f"Resultado: {soma(num1, num2)}")
        elif opcao == "2":
            print(f"Resultado: {subtracao(num1, num2)}")
        elif opcao == "3":
            print(f"Resultado: {multiplicacao(num1, num2)}")
        elif opcao == "4":
            print(f"Resultado: {divisao(num1, num2)}")

    elif opcao == "5":
        print("A encerrar a calculadora...")
    else:
        print("Opção inválida!")
```

## GRUPO B: Módulos

### 7. Módulo `my_calculator.py` e `main.py`

#### Ficheiro 1: Criar o módulo com o comando mágico Jupyter `%%writefile`

```
%%writefile my_calculator.py
def soma(a, b):
    return a + b

def subtracao(a, b):
    return a - b

def multiplicacao(a, b):
    return a * b

def divisao(a, b):
    if b == 0:
        return "Erro: Divisão por zero!"
    return a / b
```

#### Ficheiro 2: Ficheiro Principal (`main.py`)

```
%%writefile main.py
import my_calculator as calc

print("Teste do módulo my_calculator:")
print("10 + 5 =", calc.soma(10, 5))
print("10 - 5 =", calc.subtracao(10, 5))
print("10 * 5 =", calc.multiplicacao(10, 5))
print("10 / 5 =", calc.divisao(10, 5))
```

### 8. Comparação de Importações (`operacoes.py`)

#### Abordagem 1: Importar apenas uma função específica

```
from operacoes import soma

# Permite chamar a função diretamente pelo nome
resultado = soma(5, 3)
```

#### Abordagem 2: Importar o módulo completo

```
import operacoes

# Obriga a utilizar o prefixo do nome do módulo
resultado = operacoes.soma(5, 3)
```

> **Comparação:**
> * `from operacoes import soma`: Deixa o código mais curto na chamada (`soma()`), mas pode provocar conflitos se já existir outra função com o mesmo nome no código.
> * `import operacoes`: Mantém o código mais organizado e explícito (`operacoes.soma()`), evitando colisões de nomes em projetos maiores.

### 9. Utilização dos Módulos `random` e `math`

```
import random
import math

# Gerar inteiro aleatório entre 1 e 50
numero_gerado = random.randint(1, 50)

# Calcular raiz quadrada
raiz = math.sqrt(numero_gerado)

print(f"Número gerado: {numero_gerado}")
print(f"Raiz quadrada: {raiz:.2f}")
```

### 10. Módulo `calendar`

```
import calendar

def mostrar_calendario(mes, ano):
    return calendar.month(ano, mes)

def ano_bissexto(ano):
    return calendar.isleap(ano)

# Programa principal
ano = 2024
mes = 2

print(f"--- Calendário de {mes}/{ano} ---")
print(mostrar_calendario(mes, ano))

print(f"O ano {ano} é bissexto? {ano_bissexto(ano)}")
```
